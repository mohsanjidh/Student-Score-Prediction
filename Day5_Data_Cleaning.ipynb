This notebook demonstrates basic data cleaning techniques such as handling missing values, removing duplicate records, checking dataset information, and preparing the dataset for machine learning.

```python
import pandas as pd

print("Pandas imported successfully!")
```

```python
df = pd.read_csv("student_scores.csv")

print(df)
```

## Dataset Information

```python
df.info()
```

## Checking Missing Values

```python
print(df.isnull().sum())
```

## Checking Duplicate Records

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print(df)
```

```python
df.describe()
```

```python
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\nDataset is clean and ready for Machine Learning.")
```
