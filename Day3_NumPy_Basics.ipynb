## Creating NumPy Arrays

```python
import numpy as np

study_hours = np.array([1, 2, 3, 4, 5])

print(study_hours)
print(type(study_hours))
```

## Array Indexing

```python
study_hours = np.array([2, 4, 6, 8, 10])

print("First element:", study_hours[0])
print("Third element:", study_hours[2])
print("Last element:", study_hours[-1])
```

## Array Slicing

```python
study_hours = np.array([1,2,3,4,5,6,7])

print(study_hours[1:5])
print(study_hours[:4])
print(study_hours[3:])
```

## Mathematical Operations

```python
marks = np.array([60, 70, 80, 90, 100])

print("Original:", marks)

print("Add 5:", marks + 5)
print("Multiply by 2:", marks * 2)
print("Square:", marks ** 2)
```

## Array Statistics

```python
marks = np.array([60,70,80,90,100])

print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Average:", np.mean(marks))
print("Sum:", np.sum(marks))
```
