## Mini Project - Study Hours Tracker

```python
def predict_effort(hours):

    if hours < 2:
        return "Low Effort"

    elif hours < 5:
        return "Moderate Effort"

    else:
        return "High Effort"


name = input("Enter Student Name: ")

hours = int(input("Enter Study Hours: "))

result = predict_effort(hours)

print("\nStudent Name :", name)
print("Study Hours :", hours)
print("Performance :", result)
```
