This notebook demonstrates the fundamentals of data visualization using Matplotlib. It includes line charts, scatter plots, bar charts, histograms, and visual analysis of the Student Score dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

print(pd.__version__)
print("Everything works!")
```

## Load Dataset

```python
import pandas as pd

df = pd.read_csv("student_scores.csv")

print(df)
```

## Line Chart - Study Hours vs Score

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(df["StudyHours"], df["Score"], marker="o", linewidth=2)

plt.title("Study Hours vs Student Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.grid(True)

plt.show()
```

## Scatter Plot

```python
plt.figure(figsize=(8,5))

plt.scatter(df["StudyHours"], df["Score"], s=100)

plt.title("Study Hours vs Student Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.grid(True)

plt.show()
```

## Bar Chart

```python
plt.figure(figsize=(8,5))

plt.bar(df["StudyHours"], df["Score"])

plt.title("Student Scores")
plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.show()
```

## Histogram

```python
plt.figure(figsize=(8,5))

plt.hist(df["Score"], bins=5)

plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")

plt.show()
```

## Mini Project - Student Score Analysis

```python
highest = df.loc[df["Score"].idxmax()]
lowest = df.loc[df["Score"].idxmin()]

print("Highest Scoring Student")
print(highest)

print()

print("Lowest Scoring Student")
print(lowest)

print()

print("Average Score:", df["Score"].mean())
```

# Observations

- Student scores increase as study hours increase.
- The scatter plot shows a positive relationship between study hours and scores.
- The histogram helps understand the score distribution.
- Data visualization makes it easier to identify patterns before training a Machine Learning model.
